```python
import pandas as pd
import numpy as np

df=pd.read_csv("PowerQualityDistributionDataset1.csv")

df.shape
```

```python
classes = {
    1: "Normal",
    2: "3rd Harmonic Wave",
    3: "5th Harmonic Wave",
    4: "Voltage Dip",
    5: "Transient",
}

df["output"] = df["output"].map(classes)
```

```python
import matplotlib.pyplot as plt

liste=[]

fig,ax=plt.subplots(1,5,figsize=(20,4))
x_axis=[i for i in range(1,129)]

graphs=df.drop("output",axis=1)

for index,class_name in enumerate(df["output"]):
  if class_name==0 and (0 not in liste):
    ax[0].plot(x_axis,graphs.iloc[index])
    ax[0].set_title("Normal Voltage Wave")
    liste.append(0)
  elif class_name==1 and (1 not in liste):
    ax[1].plot(x_axis,graphs.iloc[index])
    ax[1].set_title("3rd Harmonic Wave")
    liste.append(1)
  elif class_name==2 and (2 not in liste):
    ax[2].plot(x_axis,graphs.iloc[index])
    ax[2].set_title("5th Harmonic Wave")
    liste.append(2)
  elif class_name==3 and (3 not in liste):
    ax[3].plot(x_axis,graphs.iloc[index])
    ax[3].set_title("Voltage Dip")
    liste.append(3)
  elif class_name==4 and (4 not in liste):
    ax[4].plot(x_axis,graphs.iloc[index])
    ax[4].set_title("Transient")
    liste.append(4)
plt.show()
```

```python
import tensorflow as tf
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Conv1D, GlobalMaxPooling1D, Dense, Dropout, MaxPooling1D
```

```python
from sklearn.model_selection import train_test_split

x=df.drop("output",axis=1)
y=df["output"]

X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
X_train,X_val,y_train,y_val=train_test_split(X_train,y_train,test_size=0.2,random_state=42,stratify=y_train)

# %64 train, %20 test, %16 validation
```

```python
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.layers import BatchNormalization

# Ortalama=0, Varyans=1

scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_val_scaled=scaler.transform(X_val)
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

X_train_scaled = np.concatenate([X_train_scaled, X_val_scaled])
y_train=np.concatenate([y_train,y_val])

svm=SVC()
svm.fit(X_train_scaled,y_train)

X_test_scaled=scaler.transform(X_test)
y_pred=svm.predict(X_test_scaled)

accuracy=accuracy_score(y_test,y_pred)
print("Support Vector Machine Accuracy:",accuracy)
```

```python
shape=(X_train.shape[1],1)
num_classes=df["output"].nunique()

model=Sequential()

model.add(Conv1D(filters=128,kernel_size=3,activation="relu",input_shape=shape))
model.add(BatchNormalization())

model.add(Conv1D(filters=64,kernel_size=5,activation="relu",input_shape=shape))
model.add(MaxPooling1D())

model.add(Conv1D(filters=128,kernel_size=5,activation="relu",input_shape=shape))
model.add(GlobalMaxPooling1D())

model.add(Dense(64,activation="relu"))
model.add(Dropout(0.5))
model.add(Dense(num_classes,activation="softmax"))

model.compile(loss="sparse_categorical_crossentropy",optimizer="adam",metrics=["accuracy"])

history=model.fit(X_train_scaled,y_train,batch_size=32,epochs=5,validation_data=(X_val_scaled,y_val))
```

```python
fig,ax=plt.subplots(1,2,figsize=(12,4))

ax[0].plot(history.history["accuracy"])
ax[0].plot(history.history["val_accuracy"])
ax[0].set_title("Doğruluk Grafikleri")
ax[0].set_xlabel("Epoch")
ax[0].set_ylabel("Accuracy")
ax[0].legend(["Train","Test"],loc="upper right")

ax[1].plot(history.history["loss"])
ax[1].plot(history.history["val_loss"])
ax[1].set_title("Kayıp Grafikleri")
ax[1].set_xlabel("Epoch")
ax[1].set_ylabel("Loss")
ax[1].legend(["Train","Test"],loc="upper right")

plt.show()
```

```python
X_test_scaled=scaler.transform(X_test)
model.evaluate(X_test_scaled,y_test)
```

```python
y_pred=np.argmax(model.predict(X_test_scaled),axis=1)
```

```python
from sklearn.metrics import confusion_matrix, classification_report

print(classification_report(y_test, y_pred))
```

```python
import seaborn as sns

cf_matrix=confusion_matrix(y_test,y_pred)

plt.figure(figsize=(6,5))
sns.heatmap(cf_matrix, annot=True, fmt="d", cmap="Blues")
plt.xlabel("Tahmin")
plt.ylabel("Gerçek Değer")
plt.title("Confusion Matrix")

plt.show()
```
